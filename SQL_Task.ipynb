Write an SQL query to display all employee details from the Employee table.

```python
import pandas as pd

df = pd.read_csv("employee_management_dataset.csv")
```

```python
df
```

Write an SQL query to display employees whose salary is greater than 50000.

```python
employees_high_salary = df[df['Salary'] > 50000]
display(employees_high_salary)
```

Write an SQL query to display employees ordered by salary in descending order.

```python
employees_sorted_by_salary = df.sort_values(by='Salary', ascending=False)
display(employees_sorted_by_salary)
```

Write an SQL query to find the maximum salary among all employees.

```python
max_salary = df['Salary'].max()
print(f"The maximum salary among all employees is: {max_salary}")
```

Write an SQL query to find the minimum salary among all employees.

```python
min_salary = df['Salary'].min()
print(f"The minimum salary among all employees is: {min_salary}")
```

Write an SQL query to calculate the average salary of employees.

```python
average_salary = df['Salary'].mean()
print(f"The average salary of employees is: {average_salary:.2f}")
```

Write an SQL query to count the total number of employees.

```python
total_employees = len(df)
print(f"The total number of employees is: {total_employees}")
```

Write an SQL query to count employees in each department.

```python
employees_by_department = df.groupby('Department').size().sort_values(ascending=False)
print("Number of employees in each department:")
display(employees_by_department)
```

Write an SQL query to display employee names along with their department manager names using INNER JOIN.

```python
# Create a copy of the DataFrame to represent managers
managers_df = df[['EmployeeID', 'FirstName', 'LastName']].copy()
managers_df.rename(columns={'EmployeeID': 'ManagerEmployeeID', 'FirstName': 'ManagerFirstName', 'LastName': 'ManagerLastName'}, inplace=True)

# Merge the original DataFrame with the managers_df
employees_with_managers = pd.merge(
    df[['EmployeeID', 'FirstName', 'LastName', 'ManagerID']],
    managers_df,
    left_on='ManagerID',
    right_on='ManagerEmployeeID',
    how='inner'
)

# Select and display the required columns
display(employees_with_managers[['FirstName', 'LastName', 'ManagerFirstName', 'ManagerLastName']])
```

Write an SQL query to display all employees and their department details using LEFT JOIN.

```python
department_details_df = employees_by_department.reset_index()
department_details_df.columns = ['Department', 'EmployeeCountInDepartment']

employees_with_department_details = pd.merge(
    df,
    department_details_df,
    on='Department',
    how='left'
)

print("Employees with Department Details (including employee count per department):")
display(employees_with_department_details[['EmployeeID', 'FirstName', 'LastName', 'Department', 'EmployeeCountInDepartment']])
```

Write an SQL query to display employees whose salary is above the average salary.

```python
employees_above_average_salary = df[df['Salary'] > average_salary]
print(f"Employees with salaries above the average salary ({average_salary:.2f}):")
display(employees_above_average_salary[['EmployeeID', 'FirstName', 'LastName', 'Salary']])
```

Write an SQL query to find departments that have employees.

```python
departments_with_employees = df['Department'].unique()
print("Departments that have employees:")
for department in departments_with_employees:
    print(f"- {department}")
```
